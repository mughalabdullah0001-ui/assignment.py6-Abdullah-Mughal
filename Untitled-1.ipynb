```python
#Q1 create a 1D Numpy array of integers from 10 to 50 (inclusive) with a step of 5. Print the array, its shape, data type, and number of dimensions.

import numpy as np

py_list = list(range(10, 51, 5))
arr = np.array(py_list)

print("Array:", arr)
print("Shape:", arr.shape)
print("Dtype:", arr.dtype)
print("Number of dimensions:", arr.ndim)
```

```python
#Q2 creat a 3x4 Numpy array of zeros, a 3x4 array of ones, a 4x4 identity matrix, and a 1D array of 8 numbers evenly spaced between 0 and 1. Print all the arrays.

zeros_arr = np.zeros((3, 4))
print("Zeros Array (3x4):\n", zeros_arr)

ones_arr = np.ones((3, 4))
print("\nOnes Array (3x4):\n", ones_arr)

identity_arr = np.eye(4)
print("\nIdentity Matrix (4x4):\n", identity_arr)

linspace_arr = np.linspace(0, 1, 8)
print("\nLinspace Array (8 numbers):\n", linspace_arr)
```

```python
#Q3 creat a 1D Numpy array of integers from 1 to 24 (inclusive). Reshape it into a 2D array of shape (4, 6) and a 3D array of shape (2, 3, 4). Print both arrays and their shapes.
arr_1d = np.arange(1, 25)
print("1D Array:", arr_1d)

arr_2d = arr_1d.reshape(4, 6)
print("\n2D Array (4, 6):\n", arr_2d)
print("Shape of 2D array:", arr_2d.shape)

arr_3d = arr_1d.reshape(2, 3, 4)
print("\n3D Array (2, 3, 4):\n", arr_3d)
print("Shape of 3D array:", arr_3d.shape)
```

```python
#Q4 given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print the first 3 elements, last 3 elements, every alternate element, and the array in reverse order.
arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])
print("First 3 elements:", arr[:3])
print("Last 3 elements:", arr[-3:])
print("Every alternate element:", arr[::2])
print("Reverse order:", arr[::-1])
```

```python
#Q5 creat a 2D array of shape (3, 4) with values from 1 to 25 (inclusive). Print the array, its shape, and the sum of all elements.

arr = np.arange(1, 26).reshape(5, 5)
print("Original Array:\n", arr)

second_row = arr[1]
print("\n(a) Second row:", second_row)

third_col = arr[:, 2]
print("(b) Third column:", third_col)

center_sub = arr[1:4, 1:4]
print("(c) Center 3x3 sub-matrix:\n", center_sub)
```

```python
#Q6 given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform basic operations (addition, subtraction, multiplication, division, and power) and calculate the dot product of the two arrays. Print the results.

import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power (a ** 2):", a ** 2)
dot_product = np.dot(a, b)
print("Dot Product:", dot_product)
```

```python
#Q7 given a = np.array ([4, 8, 2, 6, 1, 9, 3, 7, 5]), find the sum, mean min max, and standard deviation of the array. Also, find the index of the maximum and minimum values in the array. Print all the results. 




a = np.array([4, 8, 2, 6, 1, 9, 3, 7, 5])
print("Array:", a)
print("Maximum Value:", np.max(a))
print("Minimum Value:", np.min(a))
print("Index of Maximum Value:", np.argmax(a))
print("Index of Minimum Value:", np.argmin(a))
print("Mean:", np.mean(a))
print("Standard Deviation:", np.std(a))
print("Sum:", np.sum(a))
```

```python
#Q8 create an array of 20 random integers between 1 and 100 (inclusive). Reshape it into a 4x5 array. Print the array, its shape, and the maximum and minimum values along with their indices. Also, calculate the mean and standard deviation of the array. Print all the results.



arr = np.random.randint(1, 101, 20).reshape(4, 5)
print("Array:\n", arr)
print("Shape:", arr.shape)
print("Maximum Value:", np.max(arr))
print("Minimum Value:", np.min(arr))
print("Index of Maximum Value:", np.unravel_index(np.argmax(arr), arr.shape))
print("Index of Minimum Value:", np.unravel_index(np.argmin(arr), arr.shape))
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))
```

```python
#Q9 use np.where() to replace all negative values in a given 1D array with 0. Also, find the index of the maximum and minimum values in the original array. Print the modified array and the indices of the maximum and minimum values.

import numpy as np

arr = np.array([5, -3, 8, -1, 0, -7, 4])
replaced_arr = np.where(arr < 0, 0, arr)
print("Replaced Array:", replaced_arr)
max_idx = arr.argmax()
min_idx = arr.argmin()

print("Index of Maximum Value:", max_idx)
print("Index of Minimum Value:", min_idx)
```

```python
#10. Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

import numpy as np

# 1. Two 2x3 arrays
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[7, 8, 9], [10, 11, 12]])

# 2. Join them
print("Vertical:", np.vstack((a, b)))
print("Horizontal:", np.hstack((a, b)))

# 3. Sort rows and flatten
arr = np.array([[9, 3, 6], [2, 8, 1]])
result = np.sort(arr, axis=1).flatten()

print("Sorted and Flattened:", result)
```
